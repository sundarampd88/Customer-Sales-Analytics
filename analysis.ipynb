```python
# ============================================
# Customer & Sales Analytics
# Pandas + NumPy
# ============================================

import pandas as pd
import numpy as np


# 1. CSV files load karna
data1 = pd.read_csv("customer.csv")
data2 = pd.read_csv("sales.csv")





# 5. Customer + Sales merge
df = pd.merge(
    data1,
    data2,
    on="CustomerID",
    how="inner"
)





# 6. Total Sales calculate
df["Total_Sales"] = df["Quantity"] * df["Price"]



```

```python
df
```

```python
# 8. Highest Sales upar, Lowest Sales niche
df = df.sort_values(
    "Total_Sales",
    ascending=False
)




```
